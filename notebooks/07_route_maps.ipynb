# 07 Route maps
The route of a run coloured by **heart rate** and by **pace**, plus HR and pace along the distance.

```python
import sys
sys.path.append("../src")

from heatwatch import dataset
from heatwatch.analysis import routes
```

## Settings

```python
params = ["hr_5s", "gps_5s"]      # 5-s heart rate and 5-s GPS
races_only = True                  # False: all trainings
```

## Select


```python
ds = dataset.load()
sel_p, sel_t, sel_5s = dataset.select(ds.participants, ds.trainings, ds.five_s, params)
if races_only:
    sel_t = sel_t[sel_t.is_race == 1]
sel_t = sel_t[sel_t.run_id.isin(routes.usable(sel_5s, sel_t.run_id.tolist()))]   # >= 1 min of route with HR
sel_5s = sel_5s[sel_5s.run_id.isin(sel_t.run_id)]
print(f"{sel_t.study_id.nunique()} participants, {len(sel_t)} runs")
sel_t.groupby(["population", "race"]).size().unstack(fill_value=0)
```

```python
sel_t[["study_id", "population", "race", "run_id", "date", "start_time", "distance_km"]]
```

## One run
Pick a `run_id` from the table above.

```python
run_id = sel_t.run_id.iloc[8]
routes.plot(sel_t, sel_5s, run_id)
```

## Save all
One .png per run → `analysis/route_maps/`.

```python
routes.save_all(sel_t, sel_5s, sel_t.run_id.tolist())
```
